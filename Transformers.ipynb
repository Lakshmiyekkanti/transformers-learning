```python
from transformers import pipeline
classifier = pipeline('sentiment-analysis')
# text = "I love this product! It's amazing and works perfectly."
text = "I hate this product because its not good."
result = classifier(text)
print(f"Sentiment :{result}")
```

Text generation

```python
generator = pipeline('text-generation')
text = input("Enter a part of line")
result = generator(text,max_length=200)
print(result)
```

Named entity recognition

```python
ner = pipeline('ner')
text = 'Vignan University is in guntur'
result = ner(text)
print(result)
```

```python
qa = pipeline('question-answering')
text = input('Enter a part of line')
question = input('Enter a question')
result = qa(question=question,context=text)
print(result)
```

```python
summarizer = pipeline('summarization')
text = input('Enter a text')
result = summarizer(text,max_length=100,min_length=10,do_sample=False)
print(result)
```

```python
from transformers import pipeline
translator = pipeline('translation_en_to_fr')
result = translator('I love the product')
print(result)
```
